## Handling Null Values

```python
df['Size'].unique()
```

```python
df.groupby("Category")['Size'].unique()
```

```python
df.groupby("Item Purchased")['Size'].unique()
```

```python
#Treating Null Values of Electronics Category

df.loc[df['Category']=='Electronics','Size']='Not Applicable'
```

```python
#Treating Null Values of Accessories Category

df.loc[df['Category']=='Accessories','Size']='Free Size'
```

```python
#Treating Null Values of Footwear Category

df.loc[df['Category']=='Footwear','Size']='Not Available'
```

```python
#Treating Null Values of Clothing Category

df.loc[(df['Category']=='Clothing') & (df['Size'].isnull()),'Size']= df.loc[df['Category']=='Clothing']['Size'].mode()[0]
```

```python
df.groupby("Category")['Size'].unique()
```

```python
df['Review Rating'] = df['Review Rating'].fillna(df.groupby('Item Purchased')['Review Rating'].transform('mean').round(2))
```

```python
df['Previous Purchases']=df['Previous Purchases'].fillna(0)
```

```python
df.groupby('Item Purchased')['Purchase Amount (USD)'].mean().round(2)
```

```python
df['Purchase Amount (USD)'] = df['Purchase Amount (USD)'].fillna(df.groupby('Item Purchased')['Purchase Amount (USD)'].transform('mean'))
```

```python
import pandas as pd
from sqlalchemy import create_engine
import urllib

params = urllib.parse.quote_plus(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=IRON\\SQLEXPRESS;"
    "DATABASE=consumer_behavior_project;"
    "Trusted_Connection=yes;"
)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={params}"
)

with engine.connect() as conn:
    print("Connected Successfully!")
```
